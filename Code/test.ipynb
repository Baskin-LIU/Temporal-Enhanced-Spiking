```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
T = 500
dt = 0.1
time = np.arange(T)*dt
tau_V = 20
dt_tau_V = dt/tau_V
alpha = dt/50
omega = dt/6
Q = np.zeros(T)
I = np.zeros(T)
E = np.zeros(T)
V = np.zeros(T)
Q[-1] = -1

for t in range(T):
    I[t] = I[t-1] - alpha * I[t-1] - omega * Q[t-1]
    Q[t] = Q[t-1] - alpha * Q[t-1] + omega * I[t-1]
    # if t == 150:
    #     Q[t] -= 1
    V[t] = V[t-1] + dt_tau_V * (-V[t-1] + I[t-1])

plt.plot(time, Q, label='Q')
plt.plot(time, I, label='I')
plt.plot(time, V, label='V')
plt.legend()
plt.show()
```

```python
T = 1000
dt = 0.1
time = np.arange(T)*dt
tau_I, tau_V = 10, 20
dt_tau_V = dt/tau_V
dt_tau_I = dt/tau_I
I = np.zeros(T)
V = np.zeros(T)
I[-1] = 1

for t in range(T):
    I[t] = I[t-1] - dt_tau_I * I[t-1]
    V[t] = V[t-1] + dt_tau_V * (-V[t-1] + I[t-1])

plt.plot(time, I, label='I')
plt.plot(time, V, label='V')
plt.legend()
plt.show()
```

```python
T = 800
dt = 0.1
time = np.arange(T)*dt
tau_E, tau_V = 30, 20
dt_tau_V = dt/tau_V
dt_tau_E = dt/tau_E
alpha = dt/40
omega = dt/6
Q = np.zeros(T)
I = np.zeros(T)
E = np.zeros(T)
V = np.zeros(T)
Q[-1] = -1
E[-1] = 1

for t in range(T):
    Q[t] = Q[t-1] + omega * I[t-1] - alpha * Q[t-1]
    I[t] = I[t-1] - alpha * I[t-1] - omega * Q[t-1]
    E[t] = E[t-1] - dt_tau_E * (E[t-1])# + Q[t-1]-I[t-1])
    #E[t] = -Q[t-1]+I[t-1]
    V[t] = V[t-1] + dt_tau_V * (-V[t-1] + I[t-1]+ E[t-1])

plt.plot(time, Q, label='Q')
plt.plot(time, I, label='I')
plt.plot(time, E, label='E')
plt.plot(time, V, label='V')
plt.legend()
plt.show()
```
